```python
#change the day and time column to datetime format
df['day'] = pd.to_datetime(df['day'])
df['time'] = pd.to_datetime(df['time'])
```

```python
#Count of records by continent
plt.figure(figsize=(9, 5))
sns.countplot(data=df, x='continent', order=df['continent'].value_counts().index)
plt.title('Count of Records by Continent')
plt.xlabel('Continent')
plt.ylabel('Count')
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python
#Top 10 countries by total cases
top_cases = df.sort_values('Cases', ascending=False).head(10)
plt.figure(figsize=(11, 5))
sns.barplot(data=top_cases, x='country', y='Cases', palette='viridis')
plt.title('Top 10 Countries by Cases')
plt.xlabel('Country')
plt.ylabel('Cases')
plt.xticks(rotation=45, ha='right')
plt.show()
```

```python
#Relationship between tests and cases
plt.figure(figsize=(9, 5))
sns.scatterplot(data=df, x='Tests', y='Cases', hue='continent', alpha=0.7, s=60)
plt.title('Tests vs Cases')
plt.xlabel('Tests')
plt.ylabel('Cases')
plt.legend(bbox_to_anchor=(1, 1), loc='upper left')
plt.show()
```

```python
numeric_cols = ['population', 'Cases', 'Recovered', 'Deaths', 'Tests']
plt.figure(figsize=(7, 5))
sns.heatmap(df[numeric_cols].corr(), annot=True, cmap='plasma', fmt='.2f')
plt.title('Correlation Heatmap')
plt.tight_layout()
plt.show()
```
